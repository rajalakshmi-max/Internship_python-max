# **Assining grade**

```python
def grade(salary):
   if salary >= 50000:
      return 'A'
   elif salary >= 40000:
      return 'B'
   elif salary >= 30000:
      return 'C'
   else:
      return 'D'
df['Grade']=df['Salary'].apply(grade)

df[['EmployeeID','Salary','Grade']].head(20)
```

# **calculate the exp_ grade**

```python
def exp_grade(exp):
  if exp >= 15:
    return 'Senior'
  elif exp >= 8:
    return 'Mid Level'
  else:
    return 'Junior'
df['ExperienceGrade']=df['ExperienceYears'].apply(exp_grade)
df[['EmployeeID','ExperienceYears','ExperienceGrade']].head(20)
```

# **Project coding analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("project.csv")

# =========================
# BASIC ANALYSIS
# =========================

# Attrition Rate
attrition_rate = (df['Attrition'].value_counts(normalize=True) * 100)
print("Attrition Rate (%)")
print(attrition_rate)

# Department-wise Attrition
print("\nDepartment-wise Attrition")
print(pd.crosstab(df['Department'], df['Attrition']))

# =========================
# SALARY-WISE ATTRITION
# =========================

df['Salary_Group'] =pd.cut(
    df['Salary'],
    bins=[0, 5000, 10000, df['Salary'].max()],
    labels=['Low Salary', 'Medium Salary', 'High Salary']
)

print("\nSalary-wise Attrition")
print(pd.crosstab(df['Salary_Group'], df['Attrition']))

# =========================
# EXPERIENCE-WISE ATTRITION
# =========================

df['Experience_Group'] = pd.cut(
    df['ExperienceYears'],
    bins=[0, 5, 10, 20, 50],
    labels=['0-5', '6-10', '11-20', '20+']
)

print("\nExperience-wise Attrition")
print(pd.crosstab(df['Experience_Group'], df['Attrition']))

# =========================
# SATISFACTION ANALYSIS
# =========================

print("\nJob Satisfaction vs Attrition")
print(pd.crosstab(df['JobSatisfaction'], df['Attrition']))

print("\nWork Life Balance vs Attrition")
print(pd.crosstab(df['WorkLifeBalance'], df['Attrition']))

print("\nEnvironment Satisfaction vs Attrition")
print(pd.crosstab(df['EnvironmentSatisfaction'], df['Attrition']))

# =========================
# OVERTIME ANALYSIS
# =========================

print("\nOvertime vs Attrition")
print(pd.crosstab(df['OverTime'], df['Attrition']))

# =========================
# PERFORMANCE ANALYSIS
# =========================

print("\nPerformance Rating vs Attrition")
print(pd.crosstab(df['PerformanceRating'], df['Attrition']))

# =========================
# GENDER ANALYSIS
# =========================

print("\nGender-wise Attrition")
print(pd.crosstab(df['Gender'], df['Attrition']))

# =========================
# AGE GROUP ANALYSIS
# =========================

df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[20, 30, 40, 50, 100],
    labels=['20-30', '31-40', '41-50', '51+']
)

print("\nAge Group Analysis")
print(pd.crosstab(df['Age_Group'], df['Attrition']))
```

# ***Attrition by Department***

```python
import matplotlib.pyplot as plt
import pandas as pd

dept_attrition = pd.crosstab(df['Department'], df['Attrition'])

dept_attrition.plot(kind='bar')
plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

# ***Attrition by salary range***

```python
salary_attrition = pd.crosstab(df['SalaryGroup'], df['Attrition'])

salary_attrition.plot(kind='bar')
plt.title('Attrition by Salary Range')
plt.xlabel('Salary Group')
plt.ylabel('Count')
plt.show()
```

# ***Attrition by Experience***

```python
exp_attrition = pd.crosstab(df['ExperienceYears'], df['Attrition'])

exp_attrition.plot(kind='bar', figsize=(10,5))
plt.title('Attrition by Experience')
plt.xlabel('Experience Years')
plt.ylabel('Count')
plt.show()
```

# ***Attrition by Job role***

```python
job_attrition = pd.crosstab(df['JobRole'], df['Attrition'])

job_attrition.plot(kind='bar', figsize=(10,5))
plt.title('Attrition by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Count')
plt.show()
```
