CALCULATE THE TOTAL Salary

```python
import pandas as pd
df=pd.read_csv("internship_project.csv")
employee=['Department','Gender','MonthlySalary']
df['Total']=df['MonthlySalary'].sum(axis=0)
df[['EmployeeID','Department','Gender','MonthlySalary','Total']].tail(7)
```

***Assining*** ***Grade***

```python
def grade(salary):
    if salary >= 100000:
        return 'A'
    elif salary >= 80000:
        return 'B'
    elif salary >= 60000:
        return 'C'
    elif salary >= 40000:
        return 'D'
    else:
        return 'F'

df['Grade'] = df['MonthlySalary'].apply(grade)

df[['EmployeeID','MonthlySalary','Grade']].head(10)
```

```python
df['Attrition'].count()
```

```python
def exp_grade(exp):
  if exp >= 15:
        return 'Senior'
  elif  exp >= 8:
        return 'Mid Level'
  else:
        return 'Junior'
df['ExperienceGrade']=df['ExperienceYears'].apply(exp_grade)
df[['EmployeeID','ExperienceYears','ExperienceGrade']].head(20)
```

# **visualization**
# ***Bar chart***

```python
import matplotlib.pyplot as plt
import pandas as pd

dept_attrition = pd.crosstab(df['Department'], df['Attrition'])

dept_attrition.plot(kind='bar')
plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

# **Bar chart colouring**

```python
import matplotlib.pyplot as plt
import pandas as pd

dept_attrition = pd.crosstab(df['Department'], df['Attrition'])

dept_attrition.plot(kind='bar',color=['blue','purple'])
plt.title('Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

pie charts

```python
import matplotlib.pyplot as plt
import pandas as pd

dept_age= pd.crosstab(df['Department'], df['Age'])

dept_age.plot(kind='pie', subplots=True, figsize=(8, 8))
plt.title('Age by Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.show()
```

# **PIE Chart**

```python

df['Status']=df['JobSatisfaction'] .apply(lambda x:'pass' if x >=3 else 'fail')
df['Status'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title('Pass vs Fail')
plt.ylabel('')
plt.show()
```
